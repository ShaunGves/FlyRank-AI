```python
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

pages_with_ai = (df["ai_sessions_90d"] > 0).sum()
total_pages = len(df)
print(f"Pages with AI sessions: {pages_with_ai} out of {total_pages}")

avg_ai_pct = df.loc[df["ai_sessions_90d"] > 0, "ai_traffic_pct"].mean()
print(f"Average AI traffic % (for pages with AI sessions): {avg_ai_pct:.2f}%")

max_ai_sessions = df["ai_sessions_90d"].max()
print(f"Max AI sessions in 90 days for a single page: {max_ai_sessions}")
```
